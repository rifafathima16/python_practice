### DAY 4: Functions & Error Handling — def, try/except, Modules

```python
def greet(name):
  print("Hello, " + name)
  print("Good Morning!!!")

greet("Rifa")
greet("Tuba")
```

```python
def coffee_maker(sugar):
  print("Your coffee machine works!!")
  if (sugar):
    print("Your coffee has sugar")
  else:
    print("Your coffee has no sugar")
  
sugar = False
coffee_maker(sugar)
```

### Module

```python
import calculation

calculation.add(10, 20)
calculation.sub(10, 20)
calculation.mul(10, 20)
calculation.div(10, 20)
```

###  Error Handing

```python
try:
  div = 2/3
except ZeroDivisionError:
  print("Cannot Divide by Zero")
finally:
  print("program works!")
```
